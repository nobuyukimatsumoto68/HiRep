```python
import glob, re
import h5py
import numpy as np
```

```python
# locate files and sort by trajectory number
files = sorted(glob.glob("glueballs.*.h5"),
               key=lambda f: int(re.search(r'glueballs\.(\d+)\.h5', f).group(1)))
trajs = [int(re.search(r'glueballs\.(\d+)\.h5', f).group(1)) for f in files]

# read dimensions from the first file
with h5py.File(files[0], "r") as f:
    n_op = len(f.keys())
    n_t  = f["0"]["real"].shape[0]

# load all data: shape [n_traj, n_op, n_t]
real = np.empty((len(trajs), n_op, n_t))
imag = np.empty((len(trajs), n_op, n_t))

for i, fname in enumerate(files):
    with h5py.File(fname, "r") as f:
        for op in range(n_op):
            real[i, op, :] = f[str(op)]["real"][:]
            imag[i, op, :] = f[str(op)]["imag"][:]

print(f"Loaded {len(trajs)} trajectories, n_op={n_op}, n_t={n_t}")
print(f"Trajectories: {trajs}")
print(f"real shape: {real.shape}")
```
